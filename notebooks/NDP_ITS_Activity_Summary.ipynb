```python
import pandas as pd
import geopandas as gpd
```

```python
def activity_summary(activity_old_path, activity_old_lyr_name, activity_new_path, activity_new_lyr_name, output_csv_path, summary_cols):
    activity_old = gpd.read_file(activity_old_path,
                          driver='OpenFileGDB',
                          layer=activity_old_lyr_name)
    activity_new = gpd.read_file(activity_new_path,
                              driver='OpenFileGDB',
                              layer=activity_new_lyr_name)
    
    summary_old = activity_old.groupby(summary_cols).ACTIVITY_QUANTITY.sum().reset_index()
    summary_new = activity_new.groupby(summary_cols).ACTIVITY_QUANTITY.sum().reset_index()
    summary_df = summary_old.merge(summary_new, on=summary_cols, how='outer')
    summary_df = summary_df.rename({'ACTIVITY_QUANTITY_x': 'ACTIVITY_QUANTITY_old', 'ACTIVITY_QUANTITY_y': 'ACTIVITY_QUANTITY_new'}, axis=1)
    summary_df['ACTIVITY_QUANTITY_new'] = summary_df['ACTIVITY_QUANTITY_new'].fillna(0)
    summary_df['ACTIVITY_QUANTITY_old'] = summary_df['ACTIVITY_QUANTITY_old'].fillna(0)
    summary_df['ACTIVITY_QUANTITY_diff'] = summary_df['ACTIVITY_QUANTITY_new'] - summary_df['ACTIVITY_QUANTITY_old']
    summary_df['ACTIVITY_QUANTITY_abs'] = summary_df['ACTIVITY_QUANTITY_diff'].abs()
    summary_df.sort_values(by='ACTIVITY_QUANTITY_abs', ascending=False, inplace=True)
    summary_df.to_csv(output_csv_path)
    return summary_df
```

```python
summary_df = activity_summary('ITS_V2_0_Activity_webmap.gdb',
                              'ITS_V2_0_Activity',
                              'ITSV2.1_Activity_webmap.gdb',
                              'ITSV2_1_Activity_webmap',
                              'ITS_activity_v2.0_vs_v2.1_comparison.csv',
                              ['ADMINISTERING_ORG', 'ACTIVITY_DESCRIPTION', 'ACTIVITY_CAT', 'Year_txt'])
```
